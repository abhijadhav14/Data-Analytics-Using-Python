```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_date, to_timestamp
```

```python
# Create Spark Session
spark = SparkSession.builder \
    .appName("DataTypeConversionExample") \
    .getOrCreate()
```

```python
# Sample Data
data = [
    ("101", "50000.75", "2026-05-18", "2025-05-18 10:30:00"),
    ("102", "62000.50", "2026-05-19", "2025-05-19 11:45:00"),
    ("103", "45000.00", "2026-05-20", "2025-05-20 09:15:00")
]
```

```python
# Create DataFrame
df = spark.createDataFrame(
    data,
    ["emp_id", "salary", "joining_date", "login_time"]
)
```

```python
# Show Original Data
print("Original Data")
df.show()
```

```python
# Print Original Schema
print("Original Schema")
df.printSchema()
```

```python
# Data Type Conversions
df_converted = df \
    .withColumn("emp_id", col("emp_id").cast("int")) \
    .withColumn("salary", col("salary").cast("double")) \
    .withColumn("joining_date",
                to_date(col("joining_date"), "yyyy-MM-dd")) \
    .withColumn("login_time",
                to_timestamp(col("login_time"),
                             "yyyy-MM-dd HH:mm:ss"))
```

```python
# Show Converted Data
print("Converted Data")
df_converted.show()
```

```python
# Print Converted Schema
print("Converted Schema")
df_converted.printSchema()
```

```python
# Stop Spark Session
spark.stop()
```
