```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Rename, Drop, Sort columns").getOrCreate()
```

```python
data = [
    (101, "Max", 45000, "IT"),
    (102, "Oscar", 55000, "HR"),
    (103, "Lewis", 65000, "Finance")
]
```

```python
columns = ["Id", "Name", "Salary", "Department"]
```

```python
df = spark.createDataFrame(data, columns)
print("Original DataFrame")
df.show()
```

```python
df_renamed = df.withColumnRenamed("Name", "Emp_Name") \
               .withColumnRenamed("Salary", "Emp_Salary")
print("After Renaming the Columns")
df_renamed.show()
```

```python
df_dropped = df_renamed.drop("Department")
print("After Dropping Column")
df_dropped.show()
```

```python
df_muldrop = df.drop("Id", "Department")
print("After dropping multiple columns")
df_muldrop.show()
```

```python
df2 = df.toDF("E_id","E_name","E_salary","E_dept")
print("After changing the column names")
df2.show()
```

```python
df.sort("Salary").show()
```

```python
df.orderBy("Salary").show()
```

```python
df.sort(df.Salary.desc()).show()
```

```python
df.orderBy("Salary", ascending=False).show()
```

```python
df.sort("Name", "Department").show()
```
