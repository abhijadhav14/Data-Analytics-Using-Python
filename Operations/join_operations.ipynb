```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg
```

```python
spark = SparkSession.builder.master("local[*]").appName("F1 Join Operations").getOrCreate()
df = spark.read.option("header", True).option("inferSchema", True).csv("data.csv")
```

```python
drivers = df.select("driver_id", "driver_name", "team", "nationality").dropDuplicates(["driver_id"])
results = df.select("race_id", "race_date", "grand_prix", "driver_id", "finish_position", "points", "circuit")
```

```python
inner_join = results.join(drivers, on="driver_id", how="inner")
left_join = results.join(drivers, on="driver_id", how="left")
right_join = results.join(drivers, on="driver_id", how="right")
full_join = results.join(drivers, on="driver_id", how="full")
summary = inner_join.groupBy("team").agg(avg("points").alias("avg_points"))
```

```python
inner_join.show()
left_join.show()
right_join.show()
full_join.show()
summary.show()
```
