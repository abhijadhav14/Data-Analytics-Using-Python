```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import upper, lower, initcap, length, concat_ws, substring, trim
```

```python
spark = SparkSession.builder.master("local[*]").appName("F1 String Operations").getOrCreate()
df = spark.read.option("header", True).option("inferSchema", True).csv("data.csv")
```

```python
string_df = df.select(
    "driver_name",
    upper("driver_name").alias("driver_upper"),
    lower("team").alias("team_lower"),
    initcap("grand_prix").alias("grand_prix_title"),
    length("nationality").alias("nationality_length"),
    substring("circuit", 1, 10).alias("circuit_prefix"),
    trim(concat_ws(" - ", "driver_name", "team")).alias("driver_team_label")
)
```

```python
string_df.show(truncate=False)
```
