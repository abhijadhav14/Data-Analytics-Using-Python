```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import desc, asc
```

```python
spark = SparkSession.builder.master("local[*]").appName("F1 Sorting Operations").getOrCreate()
df = spark.read.option("header", True).option("inferSchema", True).csv("data.csv")
```

```python
sort_by_points = df.orderBy(desc("points"), asc("finish_position"))
sort_by_date = df.orderBy(asc("race_date"), desc("points"))
top_finishers = df.filter(df.finish_position <= 3).orderBy(asc("race_id"), asc("finish_position"))
```

```python
sort_by_points.show()
sort_by_date.show()
top_finishers.show()
```
