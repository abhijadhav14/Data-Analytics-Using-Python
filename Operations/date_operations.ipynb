```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date, year, month, dayofmonth, datediff, current_date, col, lit
```

```python
spark = SparkSession.builder.master("local[*]").appName("F1 Date Operations").getOrCreate()
df = spark.read.option("header", True).option("inferSchema", True).csv("data.csv")
```

```python
date_df = df.withColumn("race_date_parsed", to_date(col("race_date"), "yyyy-MM-dd"))
```

```python
date_operations = date_df.select(
    "grand_prix",
    "race_date_parsed",
    year("race_date_parsed").alias("race_year"),
    month("race_date_parsed").alias("race_month"),
    dayofmonth("race_date_parsed").alias("race_day"),
    datediff(current_date(), "race_date_parsed").alias("days_since_race")
)
```

```python
recent_races = date_df.filter(col("race_date_parsed") >= to_date(lit("2024-04-01"), "yyyy-MM-dd"))
```

```python
date_operations.show()
recent_races.show()
```
