```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local[*]").appName("F1 Union Operations").getOrCreate()
```

```python
df = spark.read.option("header", True).option("inferSchema", True).csv("data.csv")
first_half = df.filter(df.race_id <= 3)
second_half = df.filter(df.race_id > 3)
```

```python
simple_union = first_half.union(second_half)
union_by_name = first_half.select("driver_name", "team", "grand_prix", "points").unionByName(
    second_half.select("driver_name", "team", "grand_prix", "points")
)
```

```python
simple_union.show()
union_by_name.show()
```
