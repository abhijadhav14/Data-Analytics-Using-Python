```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder.appName("Date Functions").getOrCreate()
spark.conf.set("spark.sql.session.timeZone", "Asia/Kolkata")
```

```python
data = [ ("2025-05-20",)]
df = spark.createDataFrame(data, ["date_string"])
df.show()
```

```python
df1 = df.withColumn("date", to_date("date_string", "yyyy-MM-dd"))
df1.show()
```

```python
df1.select(current_date().alias("Today")).show()
df1.select(current_timestamp().alias("Current Time")).show(truncate=False)
df1.select("date", year("date").alias("year")).show()
df1.select("date", month("date").alias("month")).show()
df1.select("date", dayofmonth("date").alias("day")).show()
```

```python
df1.select(
    "date",
    year("date").alias("YEAR"),
    month("date").alias("MONTH"),
    dayofmonth("date").alias("DAY"),
    date_add("date", 7).alias("NEXT WEEK"),
    add_months("date", 1).alias("NEXT MONTH"),
    last_day("date").alias("MONTH END")
).show()
```
