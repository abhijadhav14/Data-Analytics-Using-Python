```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler, MinMaxScaler
```

```python
# Create Spark Session
spark = SparkSession.builder.appName("MinMaxScalerDemo").getOrCreate()
```

```python
# Sample Data
data = [(25,), (30,), (60,), (47,), (58,), (88,), (95,),(67,)]
df = spark.createDataFrame(data, ["marks"])
```

```python
# Convert column into feature vector
assembler = VectorAssembler(
    inputCols=["marks"],
    outputCol="features"
)
feature_df = assembler.transform(df)
```

```python
# Apply MinMaxScaler
scaler = MinMaxScaler(
    inputCol="features",
    outputCol="scaledFeatures"
)
```

```python
model = scaler.fit(feature_df)
scaled_df = model.transform(feature_df)
```

```python
scaled_df.select("marks", "scaledFeatures").show()
```
