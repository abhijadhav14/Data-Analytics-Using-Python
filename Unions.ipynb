```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("SetOperationsDemo").getOrCreate()
```

```python
data1 = [
    (101, "Max", 50009),
    (102, "Lewis", 60000),
    (103, "Charles", 55000),
    (104, "Liam", 65000)]
data2 = [
    (103, "Oscar", 55000),
    (104, "Issac", 65000),
    (105, "Nico", 70000),
    (106, "Carlos", 62000),
    (102, "Fernando", 60000)
]

df1 = spark.createDataFrame(data1,["EmpID", "EmpName", "Salary"])
df2 = spark.createDataFrame(data2,["EmpID", "EmpName", "Salary"])
df1.show()
df2.show()
```

```python
union_df = df1.union(df2)

union_df.show()
```

```python
data3 = [
    ("Gabrial", 107, 75000),
    ("Daniel", 108, 80000)
]
df3 = spark.createDataFrame(data3,["EmpName", "EmpID", "Salary"])
df3.show()
union_name_df = df1.unionByName(df3)
union_name_df.show()
```

```python
data4 = [(109, "George", "IT"),(110, "Kimi", "HR")]
df4 = spark.createDataFrame(data4,["EmpID", "EmpName", "Department"])
df4.show()
```

```python
union_missing_df = df1.unionByName(df4,allowMissingColumns=True)
union_missing_df.show()
```

```python
intersect_df = df1.intersect(df2)
intersect_df.show()
```

```python
subtract_df1 = df1.subtract(df2)
subtract_df1.show()
```

```python
subtract_df2 = df2.subtract(df1)
subtract_df2.show()
```

```python
spark.stop()
```
