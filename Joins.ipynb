```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Joins").getOrCreate()
```

```python
employee_data = [
    (1, "Max", 101),
    (2, "Issac", 102),
    (3, "Nico", 103),
    (4, "Fernando", 104),
    (5, "Charles", None)
]
employee_columns = ["emp_id", "emp_name", "dept_id"]
employee_df = spark.createDataFrame(employee_data, employee_columns)
print("===== Employee Data =====")
employee_df.show()
```

```python
department_data = [
    (101, "HR"),
    (102, "Finance"),
    (103, "IT"),
    (105, "Marketing")
]
department_columns = ["dept_id", "dept_name"]
department_df = spark.createDataFrame(department_data, department_columns)
print("===== Department Data =====")
department_df.show()
```

```python
print("===== INNER JOIN =====")
inner_join_df = employee_df.join(department_df,employee_df.dept_id == department_df.dept_id,"inner")
inner_join_df.show()
```

```python
print("===== LEFT JOIN =====")
left_join_df = employee_df.join(
    department_df,
    employee_df.dept_id == department_df.dept_id,
    "left"
)
left_join_df.show()
```

```python
print("===== RIGHT JOIN =====")
right_join_df = employee_df.join(
    department_df,
    employee_df.dept_id == department_df.dept_id,
    "right"
)
right_join_df.show()
```

```python
print("===== FULL OUTER JOIN =====")
outer_join_df = employee_df.join(
    department_df,
    employee_df.dept_id == department_df.dept_id,
    "outer"
)
outer_join_df.show()
```

```python
spark.stop()
```
