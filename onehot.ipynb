```python
from pyspark.sql import SparkSession
from pyspark.ml import *
from pyspark.ml.feature import StringIndexer
from pyspark.ml.feature import OneHotEncoder
```

```python
spark = SparkSession.builder \
    .appName("OneHotEncoderDemo") \
    .getOrCreate()
```

```python
data = [("Red",),
        ("Blue",),
        ("Green",),
        ("Red",),
        ("Blue",)]
```

```python
df = spark.createDataFrame(data, ["Color"])
```

```python
print("Original Data:")
df.show()
```

```python
indexer = StringIndexer(inputCol="Color", outputCol="ColorIndex")
indexed_df = indexer.fit(df).transform(df)
print("After String Indexing:")
indexed_df.show()
```

```python
encoder = OneHotEncoder(inputCols=["ColorIndex"],
    outputCols=["ColorVector"]
)
encoded_df = encoder.fit(indexed_df).transform(indexed_df)
print("After One-Hot Encoding:")
encoded_df.show(truncate=False)

spark.stop()
```
