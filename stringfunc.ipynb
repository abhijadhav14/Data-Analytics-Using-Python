```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder.appName("AllStringFunctions").getOrCreate()
data = [
    ("Neymar Jr", " BRAZIL "),
    ("C Ronaldo", " pORTUGAL "),
    ("LEO Messi", " ARGENTINA ")
]
```

```python
# Create DataFrame
df = spark.createDataFrame(data, ["Name", "Country"])
df.show()
```

```python
# Apply all string functions
result = df.select(

    # 1. upper()
    upper("Name").alias("upper_name"),

    # 2. lower()
    lower("Name").alias("lower_name"),

    # 3. initcap()
    initcap("Country").alias("initcap_country"),

    # 4. trim()
    trim("Country").alias("trim_country"),

    # 5. ltrim()
    ltrim("Country").alias("left_trim"),

    # 6. rtrim()
    rtrim("Country").alias("right_trim"),

    # 7. length()
    length("Name").alias("name_length"),

    # 8. concat()
    concat(col("Name"), lit(" - "), col("Country")).alias("concat_value"),

    # 9. substring()
    substring("Name", 1, 6).alias("substring_name"),

    # 10. regexp_replace()
    regexp_replace("Country", "ARGENTINA", "Argentina").alias("replaced_country"),

    # 11. split()
    split("Name", " ").alias("split_name"),

    # 12. contains()
    col("Name").contains("Messi").alias("contains_Messi"),

    # 13. startswith()
    col("Name").startswith("N").alias("starts_with_N"),

    # 14. endswith()
    col("Name").endswith("o").alias("ends_with_o"),

    # 15. instr()
    instr("Name", "Messi").alias("position_Messi"),

    # 16. repeat()
    repeat("Name", 2).alias("repeat_name"),

    # 17. reverse()
    reverse("Name").alias("reverse_name"),

)

# Display Result
result.show(truncate=False)
```
